```python
def Calcular(p, h):
    imc = p / (h * h)
    return imc
    
```

```python
def C_imc(imc):
    if imc < 18.5:
        return "Abaixo do peso!"
    elif imc < 25:
        return "Peso normal!"
    elif imc < 30:
        return "Sobrepeso!"
    else:
        return "Obesidade!"
```

```python
h = float(input("Altura em M: "))
p = float(input("Peso em kg: "))
imc = Calcular(p, h)
print(f"imc: {imc:.2f} - {C_imc(imc)}")
```

```python
def eh_primo(n):
    if n <= 1:
        return False
    else:
      for i in range (2,n):
          if (n % i) == 0:
              return False
      return True 
```

```python
n = int(input("Digite um número: "))
if eh_primo(n):
    print("O número é primo!")
else:
    print("O número não é primo!")
```
