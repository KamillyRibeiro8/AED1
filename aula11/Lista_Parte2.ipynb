### 1- Faça um programa que peça uma nota, entre zero e dez. Mostre uma mensagem caso o valor seja inválido e continue pedindo até que o usuário informe um valor válido.

```python

nota = -1
while nota < 0 or nota > 10:
    nota = int(input("Informe uma nota (0 a 10): "))
    if nota < 0 or nota > 10:
        print("Valor inválido!")

print(f"Valor válido: {nota}")
```

### 2 - Faça um programa que leia um nome de usuário e a sua senha e não aceite a senha igual ao nome do usuário, mostrando uma mensagem de erro e voltando a pedir as informações

```python

nome = input("Digite o nome de usuário: ")
senha = input("Digite a senha: ")

while senha == nome:
    print("Erro: A senha não pode ser igual ao nome de usuário. Tente novamente.")
    senha = input("Digite a senha: ")

print("Usuário e senha cadastrados com sucesso")
```

### 3 -  Faça um programa que leia e valide as seguintes informações:
##### a. Nome: maior que 3 caracteres;
##### b. Idade: entre 0 e 150;
##### c. Salário: maior que zero;
##### d. Sexo: 'f' ou 'm';
##### e. Estado Civil: 's', 'c', 'v', 'd';

```python
print("DIGITE AS SEGUINTES INFORMAÇÕES:") # Corrigido: 'print' minúsculo

nome = input("Nome: ")
while len(nome) <= 3:
    print("Erro: O nome deve ter mais de 3 caracteres.")
    nome = input("Nome: ")

idade = int(input("Idade: "))
while idade < 0 or idade > 150:
    print("Erro: A idade deve estar entre 0 e 150.")
    idade = int(input("Idade: "))

salario = float(input("Salário: "))
while salario <= 0:
    print("Erro: O salário deve ser maior que zero.")
    salario = float(input("Salário: "))

sexo = input("Sexo ('f' ou 'm'): ").lower()
while sexo not in ['f', 'm']:
    print("Erro: Sexo deve ser 'f' ou 'm'.")
    sexo = input("Sexo: ").lower()

estado_civil = input("Estado Civil ('s', 'c', 'v', 'd'): ").lower()
while estado_civil not in ['s', 'c', 'v', 'd']:
    print("Erro: Estado civil deve ser 's', 'c', 'v' ou 'd'.")
    estado_civil = input("Estado Civil: ").lower()

print("Todas as informações foram validadas com sucesso!")

```
