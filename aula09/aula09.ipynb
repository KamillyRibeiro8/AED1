```python
numero = int(input("Digite um número:"))
if numero % 5 == 0 and numero % 3 == 0:
    print("O número é divisível por 5 e por 3 ao mesmo tempo.")
else:
    print("O número não atende aos critérios.")
```

```python
valor_compra = float(input("Digite o valor da compra: R$ "))

if valor_compra <= 1000:
    desconto = 0.10
elif valor_compra <= 5000:
    desconto = 0.20
else:
    desconto = 0.30

valor_final = valor_compra * (1 - desconto)
print(f"Desconto aplicado: {desconto*100:.0f}%")
print(f"Valor total com desconto: R$ {valor_final:.2f}")
```

```python
largura = float(input("Largura do aposento (m): "))
comprimento = float(input("Comprimento do aposento (m): "))
capacidade_lata = float(input("Quantidade de litros da lata de tinta: "))

pe_direito = 2.80
area_porta = 0.80 * 2.10

area_paredes = 2 * (largura * pe_direito) + 2 * (comprimento * pe_direito)
area_liquida = area_paredes - area_porta

litros_necessarios = area_liquida / 3
latas_necessarias = litros_necessarios / capacidade_lata

print(f"\nÁrea a ser pintada: {area_liquida:.2f} m²")
print(f"Litros de tinta necessários: {litros_necessarios:.2f} L")
print(f"Quantidade de latas: {latas_necessarias:.2f}")
```
