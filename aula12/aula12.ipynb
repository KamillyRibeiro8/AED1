```python
cidades = []
while True:
    cidade = input("Digite o nome da cidade ,<Para finalizar digite 'sair'>")

    if cidade.lower() == 'sair':
        break
    cidades.append(cidade)

print("######################################")
print("########  Cidades Digitadas   ########")
cidades.sort()
for cidade in cidades:
    print(cidades)
```
