```python
norte = ['AM', 'PA', 'AP', 'TO', 'RR', 'RO', 'AC']
nordeste = ['BA', 'PE', 'CE', 'RN', 'PB', 'MA', 'PI', 'AL', 'SE']
centro_oeste = ['DF', 'GO', 'MT', 'MS']
sul = ['PR', 'RS', 'SC']
sudeste = ['SP', 'RJ', 'MG', 'ES']
    


def V_servicos(peso, estado, presente=False):

    
    estado_formatado = estado.strip().upper()

   
    if estado_formatado in norte:
        custo_regiao = 50.00
    elif estado_formatado in nordeste:
        custo_regiao = 50.00
    elif estado_formatado in centro_oeste:
        custo_regiao = 25.00
    elif estado_formatado in sul:
        custo_regiao = 25.00
    elif estado_formatado in sudeste:
        custo_regiao = 0.00
    else:
        raise ValueError(f"Estado '{estado}' inválido.")

    
    if peso <= 0.500:      
        C_peso = 15.00
    elif peso <= 2.00:      
        C_peso = 43.00 
    else:                  
        C_peso = 103.00

   

    if presente:
        C_presente = 10.00
    else:
        C_presente = 0.00

    
    V_total = C_peso + custo_regiao + C_presente
    
    return V_total
```

```python
peso = float(input("digite o peso em kg: "))
estado = input("digite a sigla do estado: ")
presente = input("é presente? (s/n): ").strip().lower() == 's'

print(f"Peso:{peso}")
print(f"Estado:{estado}")
print(f"Presente?:{presente}")

valor_final = V_servicos(peso, estado, presente)
print(f"Valor Total do Serviço: R$ {valor_final:.2f}")
```
